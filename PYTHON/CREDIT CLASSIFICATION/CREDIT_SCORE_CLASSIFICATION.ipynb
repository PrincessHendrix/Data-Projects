1. Which occupations have the best credit profiles?

```python
# Define a mapping for Credit_Mix to numerical values
credit_mix_mapping = {
    'Good': 2,
    'Standard': 1,
    'Bad': 0,
    'Unknown': np.nan # Treat 'Unknown' as NaN for mean calculation
}
```

```python
# Apply the mapping to create a new numerical column
df['Credit_Mix_Score'] = df['Credit_Mix'].map(credit_mix_mapping)

# Group by 'Occupation' and calculate the mean of the new score
# Drop NaN values (from 'Unknown' credit mix) when calculating the mean
occupation_credit_profile = df.groupby('Occupation')['Credit_Mix_Score'].mean().sort_values(ascending=False)
print(occupation_credit_profile)
```

```python
print("Occupations with the best credit profiles based on Credit_Mix_Score (higher is better):")
display(occupation_credit_profile)
```

How does age group affect borrowing behavior?

```python
# Create age groups from continuous age data
# age_bins=[0, 25, 35, 45, 55, 65, 75, 75+] defines the age ranges:["0-24", "25-34", "35-44", "45-54", "55-64", "65-74", "75+"]
#age_bins = [0, 25, 35, 45, 55, 65, 75, 100]
#age_labels = ["0-24", "25-34", "35-44", "45-54", "55-64", "65-74", "75+"]
#code By default, pd.cut is right=True, which means a bin of [0, 25] includes 25. Changing it to right=False makes the bin [0, 25),
#meaning it includes 0 up to 24.99, but excludes 25.
df['Age_Group'] = pd.cut(
                         df['Age'],
                         bins = [1, 25, 35, 45, 55, 65, 120],
                         labels = ["1-24", "25-34", "35-44", "45-54", "55-64", "65+"],
                         right=False
)
```

```python
# Borrowing behavior: average loan amount per age group
avg_loan_by_age = df.groupby("Age_Group")["Num_of_Loan"].mean()
customer_count = df.groupby("Age_Group")["Customer_ID"].nunique()
avg_credit_score = df.groupby("Age_Group")["Credit_Mix_Score"].mean()
```

```python
# Returns a single integer of unique customers
total_distinct_customers = df['Customer_ID'].nunique()

print(f"Total Unique Customers: {total_distinct_customers}")
```

```python
#TO GET DISTINCT CUSTOMER COUNT BY AGE RANGE, RECALL THAT CUSTOMER_COUNT HAS BEEN DERIVED TO BE OF UNIQUE COUNTING WITH nunique()
print(customer_count)
```

```python
age_profile_summary = pd.DataFrame({
    'Average_Num_of_Loans': avg_loan_by_age,
    'Customer_Count': customer_count,
    'Average_Credit_Score': avg_credit_score
})
print("Age Group Profile:")
display(age_profile_summary)
```

Which income band carries the most debt?

```python
#qcut is a pandas function that divides data into quantile-based bins.q=4 means we want to split the income data into 4 equal-sized groups (quartiles).
#Each group will contain roughly 25% of the customers, based on their income distribution.
df['Income_Band'] = pd.qcut(
    df['Annual_Income'], q=4,
    labels=['low level income', 'mid level income', 'High level income', 'Very High Level Income']
)
```

```python
# Aggregate debt metrics by income band
annualincome_vs_debt =  df.groupby ('Income_Band')["Outstanding_Debt"].mean()
annualincome_vs_total_debt = df.groupby ('Income_Band')["Outstanding_Debt"] .sum()
customer_count= df.groupby ('Income_Band')["Customer_ID"].nunique()
monthlyincome_vs_debt = df.groupby ('Income_Band') ['Monthly_Inhand_Salary'].mean()
```

```python
Debt_by_income_band = df.groupby("Income_Band").agg(
    avg_debt=("Outstanding_Debt", "mean"),
    total_debt=("Outstanding_Debt", "sum"),
    customer_count=("Customer_ID", "count"),
    avg_monthly_salary=("Monthly_Inhand_Salary", "mean")
)
Debt_by_income_band = Debt_by_income_band.round(2)
print(Debt_by_income_band)
```

What credit mix predicts default risk?

```python
risk_profile = df.groupby("Credit_Mix").agg(
                   customer_count=("Customer_ID", "count"),
                   Avg_Delayed_Payments  = ('Num_of_Delayed_Payment', 'mean'),
                   Avg_Delay_Days        = ('Delay_from_due_date', 'mean'),
                   Avg_Outstanding_Debt  = ('Outstanding_Debt', 'mean'),
                   Avg_Num_Loans         = ('Num_of_Loan', 'mean')
                   )
risk_profile = risk_profile.round(2)
print(risk_profile)
```

How does credit history length affect risk?

```python
#np.inf is NumPy’s representation of infinity. the last bin is [400, ∞), meaning any credit history age of 400 months or more will fall into the "400+ Mnths" category.
df['History_band'] = pd.cut(
    df['Credit_History_Age_Months'],
    bins=[0, 60, 120, 200, 400, np.inf],
    labels=['0–59 Mnths', '60–119 Mnths', '120–199 Mnths', '200-399 Mnths', '400+ Mnths'],
    right=False
)
```

```python
# Aggregate risk metrics by history band
history_length_risk = df.groupby("History_band").agg(
    avg_num_of_delayed_payment=("Num_of_Delayed_Payment", "mean"),              # proportion of customers who defaulted
    avg_credit_score=("Credit_Mix_Score", "mean"), # average credit score
    avg_debt=("Outstanding_Debt", "mean"),         # average debt
    customer_count=("Customer_ID", "count")        # number of customers
)
print(history_length_risk)
```
